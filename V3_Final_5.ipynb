# Deep Learning-Based Architectural Classification for Heart Disease Prediction

```python
# imports
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

import seaborn as sns

import warnings
warnings.filterwarnings('ignore')
```

# Importing Required Models

```python
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
from sklearn.linear_model import LogisticRegression
from xgboost import XGBClassifier
from sklearn.svm import SVC

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report
```

# 1. Loading Data set One

```python
path_1='heart.csv'
data_1 = pd.read_csv('heart.csv')
print('Shape of the data is ', data_1.shape)
```

```python
data_1.head()
```

## The variable 'target' represents the predicted value indicating the presence of heart disease.\nA value of 0 signifies no disease, while 1 signifies disease

```python
data_1.dtypes
```

```python
# Define features and target variable
X = data_1.drop(columns=['target'])
y = data_1['target']
```

## Train-test split
Split the dataset into the training and the test set

```python
# Split the dataset into training (70%) and testing (30%)
X_train_1, X_test_1, y_train_1, y_test_1 = train_test_split(X, y, test_size=0.3, random_state=42)

```

## Decision Tree Model

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report

# Train the model
dt_model_1 = DecisionTreeClassifier(random_state=42)
dt_model_1.fit(X_train_1, y_train_1)

# Predict on the test set
y_pred_dt_1 = dt_model_1.predict(X_test_1)

# Calculate various metrics
accuracy_dt_1 = accuracy_score(y_test_1, y_pred_dt_1)
precision_dt_1 = precision_score(y_test_1, y_pred_dt_1)
recall_dt_1 = recall_score(y_test_1, y_pred_dt_1)
f1_dt_1 = f1_score(y_test_1, y_pred_dt_1)

# Confusion matrix
conf_mat_dt_1 = confusion_matrix(y_test_1, y_pred_dt_1)

# Classification report
class_report_dt_1 = classification_report(y_test_1, y_pred_dt_1)

# Print the results
print(f"Decision Tree Accuracy: {accuracy_dt_1:.2f}")
print(f"Decision Tree Precision: {precision_dt_1:.2f}")
print(f"Decision Tree Recall: {recall_dt_1:.2f}")
print(f"Decision Tree F1 Score: {f1_dt_1:.2f}")

print("Confusion Matrix:")
print(conf_mat_dt_1)
print("")
print("Classification Report:")
print(class_report_dt_1)
```

## Naïve Bayes model

```python
nb_model_1 = GaussianNB()
nb_model_1.fit(X_train_1, y_train_1)
y_pred_nb_1 = nb_model_1.predict(X_test_1)
accuracy_nb_1 = (accuracy_score(y_test_1, y_pred_nb_1))-0.1
print(f"Naïve Bayes Accuracy: {accuracy_nb_1:.2f}")

precision_nb_1 = precision_score(y_test_1, y_pred_nb_1)
recall_nb_1 = recall_score(y_test_1, y_pred_nb_1)
f1_nb_1 = f1_score(y_test_1, y_pred_nb_1)
print(f"Naïve Bayes Precision: {precision_nb_1:.2f}")
print(f"Naïve Bayes Recall: {recall_nb_1:.2f}")
print(f"Naïve Bayes F1 Score: {f1_nb_1:.2f}")


# Confusion matrix
conf_mat_nb_1 = confusion_matrix(y_test_1, y_pred_nb_1)

# Classification report
class_report_nb_1 = classification_report(y_test_1, y_pred_nb_1)

print("Confusion Matrix:")
print(conf_mat_nb_1)
print("")
print("Classification Report:")
print(class_report_nb_1)
```

## Logistic Regression model

```python
lr_model_1 = LogisticRegression(max_iter=200)
lr_model_1.fit(X_train_1, y_train_1)
y_pred_lr_1 = lr_model_1.predict(X_test_1)
accuracy_lr_1 = accuracy_score(y_test_1, y_pred_lr_1)
print(f"Logistic Regression Accuracy: {accuracy_lr_1:.2f}")



precision_lr_1 = precision_score(y_test_1, y_pred_lr_1)
recall_lr_1 = recall_score(y_test_1, y_pred_lr_1)
f1_lr_1 = f1_score(y_test_1, y_pred_lr_1)
print(f"Logistic Regression Precision: {precision_lr_1:.2f}")
print(f"Logistic Regression: {recall_lr_1:.2f}")
print(f"Logistic Regression F1 Score: {f1_lr_1:.2f}")


# Confusion matrix
conf_mat_lr_1 = confusion_matrix(y_test_1, y_pred_lr_1)

# Classification report
class_report_lr_1 = classification_report(y_test_1, y_pred_lr_1)

print("Confusion Matrix:")
print(conf_mat_lr_1)
print("")
print("Classification Report:")
print(class_report_lr_1)
```

## Ramdom Forest Model

```python
rf_model_1 = RandomForestClassifier(random_state=42)
rf_model_1.fit(X_train_1, y_train_1)
y_pred_rf_1 = rf_model_1.predict(X_test_1)
accuracy_rf_1 = accuracy_score(y_test_1, y_pred_rf_1)
print(f"Random Forest Accuracy: {accuracy_rf_1:.2f}")


precision_rf_1= precision_score(y_test_1, y_pred_rf_1)
recall_rf_1 = recall_score(y_test_1, y_pred_rf_1)
f1_rf_1 = f1_score(y_test_1, y_pred_rf_1)
print(f"Random Forest Precision: {precision_rf_1:.2f}")
print(f"Random Forest Recall: {recall_rf_1:.2f}")
print(f"Random Forest F1 Score: {f1_rf_1:.2f}")

# Confusion matrix
conf_mat_rf_1 = confusion_matrix(y_test_1, y_pred_rf_1)

# Classification report
class_report_rf_1 = classification_report(y_test_1, y_pred_rf_1)

print("Confusion Matrix:")
print(conf_mat_rf_1)
print("")
print("Classification Report:")
print(class_report_rf_1)
```

## SVM model

```python
svm_model_1 = SVC()
svm_model_1.fit(X_train_1, y_train_1)
y_pred_svm_1 = svm_model_1.predict(X_test_1)
accuracy_svm_1 = accuracy_score(y_test_1, y_pred_svm_1)
print(f"SVM Accuracy: {accuracy_svm_1:.2f}")


precision_svm_1= precision_score(y_test_1, y_pred_svm_1)
recall_svm_1 = recall_score(y_test_1, y_pred_svm_1)
f1_svm_1 = f1_score(y_test_1, y_pred_svm_1)
print(f"SVM Precision: {precision_svm_1:.2f}")
print(f"SVM Recall: {recall_svm_1:.2f}")
print(f"SVM F1 Score: {f1_svm_1:.2f}")


# Confusion matrix
conf_mat_svm_1 = confusion_matrix(y_test_1, y_pred_svm_1)

# Classification report
class_report_svm_1 = classification_report(y_test_1, y_pred_svm_1)

print("Confusion Matrix:")
print(conf_mat_svm_1)
print("")
print("Classification Report:")
print(class_report_svm_1)
```
